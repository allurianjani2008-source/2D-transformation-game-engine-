```python
import numpy as np
import matplotlib.pyplot as plt

print("Libraries loaded successfully!")
```

```python
import numpy as np

# A simple square
square = np.array([
    [0, 0, 1],
    [2, 0, 1],
    [2, 2, 1],
    [0, 2, 1],
    [0, 0, 1]
])

print("Original Square:")
print(square)
```

```python
import numpy as np

def translation(tx, ty):
    return np.array([
        [1, 0, tx],
        [0, 1, ty],
        [0, 0, 1]
    ])


def scaling(sx, sy):
    return np.array([
        [sx, 0, 0],
        [0, sy, 0],
        [0, 0, 1]
    ])


def rotation(angle):
    radians = np.radians(angle)

    return np.array([
        [np.cos(radians), -np.sin(radians), 0],
        [np.sin(radians),  np.cos(radians), 0],
        [0, 0, 1]
    ])

print("Transformation functions created!")
```

```python
tx = 2       # Translation in X direction
ty = 1       # Translation in Y direction

sx = 1.5     # Scaling in X direction
sy = 1.5     # Scaling in Y direction

angle = 30   # Rotation angle in degrees

print("Translation X:", tx)
print("Translation Y:", ty)
print("Scaling X:", sx)
print("Scaling Y:", sy)
print("Rotation Angle:", angle, "degrees")
```

```python
import numpy as np

# Translation Matrix
T = np.array([
    [1, 0, tx],
    [0, 1, ty],
    [0, 0, 1]
])

# Scaling Matrix
S = np.array([
    [sx, 0, 0],
    [0, sy, 0],
    [0, 0, 1]
])

# Rotation Matrix
r = np.radians(angle)

R = np.array([
    [np.cos(r), -np.sin(r), 0],
    [np.sin(r),  np.cos(r), 0],
    [0, 0, 1]
])

# Combine matrices
M = T @ R @ S

print("Translation Matrix:")
print(T)

print("\nScaling Matrix:")
print(S)

print("\nRotation Matrix:")
print(np.round(R, 3))

print("\nFinal Transformation Matrix:")
print(np.round(M, 3))
```

```python
transformed_square = (M @ square.T).T

print("Original Coordinates:")
print(square[:, :2])

print("\nNew Coordinates:")
print(np.round(transformed_square[:, :2], 2))
```

```python
plt.figure(figsize=(8, 8))

# Original square
plt.plot(
    square[:, 0],
    square[:, 1],
    marker="o",
    label="Original Object"
)

# Transformed square
plt.plot(
    transformed_square[:, 0],
    transformed_square[:, 1],
    marker="o",
    label="Transformed Object"
)

# X and Y axes
plt.axhline(0)
plt.axvline(0)

plt.grid(True)
plt.axis("equal")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("2D Transformation Game Engine")

plt.legend()
plt.show()
```
